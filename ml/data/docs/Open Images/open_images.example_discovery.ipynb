# Open Images v7 Exploration via FiftyOne

Quick exploration of Google Open Images v7 dataset.

## Dataset Overview & Schema

- **Source Used for Analysis:** FiftyOne Dataset Zoo (`open-images-v7`, validation split sample)
- **Where to Get Full Dataset Later:** Google Open Images portal [https://storage.googleapis.com/openimages/web/index.html](https://storage.googleapis.com/openimages/web/index.html) or via FiftyOne `foz.load_zoo_dataset("open-images-v7")`
- **Full Dataset Size:** ~500 GB – 1 TB for full 9M images; validation split alone is ~12 GB (41,620 images); segmentation masks zip is ~10 GB
- **Classes:** 350+ object classes with instance segmentations and bounding boxes

### FiftyOne Dataset Schema
| Field | Type | Description |
| :--- | :--- | :--- |
| `id` | `str` | Unique sample ID. |
| `filepath` | `str` | Local path to the image file. |
| `ground_truth` | `fo.Detections` | Bounding boxes and labels for object detection. |
| `segmentations` | `fo.Detections` | Instance segmentation masks with polygon / bitmap data. |
| `positive_labels` | `fo.Classifications` | Image-level labels confirmed present. |
| `negative_labels` | `fo.Classifications` | Image-level labels confirmed absent. |


### 1. Load sample and launch FiftyOne visualizer


```python
import fiftyone as fo
import fiftyone.zoo as foz

# Load a sample of Open Images v7 validation set using FiftyOne Zoo
dataset = foz.load_zoo_dataset(
    "open-images-v7",
    split="validation",
    max_samples=10,
    shuffle=True,
    seed=42,
)

# Inspect fields and launch interactive FiftyOne visualizer
session = fo.launch_app(dataset, auto=False)
```
