# COCO 2017 Segmentation Exploration via FiftyOne

Interactive visualization of COCO instance segmentations filtered by class.

## Dataset Overview & Schema

- **Source Used for Analysis:** FiftyOne Dataset Zoo (`coco-2017`, validation split filtered for `person`, `car`)
- **Where to Get Full Dataset Later:** [https://cocodataset.org/#download](https://cocodataset.org/#download) or via FiftyOne: `foz.load_zoo_dataset("coco-2017")`
- **Full Dataset Size:** ~27 GB compressed (~118k train + 5k val + 41k test images; annotations ~241 MB)
- **Classes:** 80 object categories

### FiftyOne Dataset Schema
| Field | Type | Description |
| :--- | :--- | :--- |
| `filepath` | `str` | Local path to the COCO image file. |
| `ground_truth` | `fo.Detections` | Detection bounding boxes and class labels. |
| `segmentations` | `fo.Detections` | Instance masks with polygon/RLE boundaries for filtered classes. |


### 1. Load COCO samples with segmentations


```python
import fiftyone as fo
import fiftyone.zoo as foz

# Load COCO 2017 validation samples filtered by specific classes
dataset = foz.load_zoo_dataset(
    "coco-2017",
    split="validation",
    label_types=["detections", "segmentations"],
    classes=["person", "car"],
    max_samples=50,
)

# Launch FiftyOne interactive app to inspect segmentation masks
session = fo.launch_app(dataset)
```
